# Exercise 1: Rapidity and Pseudorapidity

## Part a) Transformation of Rapidity under a Lorentz Boost

The rapidity $y$ of a particle is defined as:
$$y = \frac{1}{2} \ln \left( \frac{E + p_z}{E - p_z} \right)$$

We want to show that under a Lorentz boost with velocity $\beta$ along the $+z$ direction:
$$y \xrightarrow{\beta} y' = y - \tanh^{-1}\beta$$

---

### Step 1: Standard Lorentz Transformation for $E$ and $p_z$

For a boost $\beta$ along the $z$-axis (in natural units $c = 1$):
$$E' = \gamma (E - \beta p_z)$$
$$p_z' = \gamma (p_z - \beta E)$$

where $\gamma = \frac{1}{\sqrt{1 - \beta^2}}$.

---

### Step 2: Linear Combinations $(E' + p_z')$ and $(E' - p_z')$

1. **Sum:**
   $$E' + p_z' = \gamma (E - \beta p_z + p_z - \beta E) = \gamma \big[ E(1 - \beta) + p_z(1 - \beta) \big] = \gamma (1 - \beta)(E + p_z)$$

2. **Difference:**
   $$E' - p_z' = \gamma (E - \beta p_z - p_z + \beta E) = \gamma \big[ E(1 + \beta) - p_z(1 + \beta) \big] = \gamma (1 + \beta)(E - p_z)$$

---

### Step 3: Ratio and Logarithm

Taking the ratio eliminates $\gamma$:
$$\frac{E' + p_z'}{E' - p_z'} = \frac{\gamma (1 - \beta)(E + p_z)}{\gamma (1 + \beta)(E - p_z)} = \left( \frac{1 - \beta}{1 + \beta} \right) \left( \frac{E + p_z}{E - p_z} \right)$$

Taking $\frac{1}{2}\ln(\dots)$ of both sides:
$$y' = \frac{1}{2} \ln \left( \frac{E' + p_z'}{E' - p_z'} \right) = \underbrace{\frac{1}{2} \ln \left( \frac{E + p_z}{E - p_z} \right)}_{\text{initial rapidity } y} + \frac{1}{2} \ln \left( \frac{1 - \beta}{1 + \beta} \right)$$

$$y' = y + \frac{1}{2} \ln \left( \frac{1 - \beta}{1 + \beta} \right)$$

---

### Step 4: Relation to $\tanh^{-1}\beta$

Using the hint for hyperbolic tangent:
$$\tanh x = \frac{e^x - e^{-x}}{e^x + e^{-x}} = \frac{e^{2x} - 1}{e^{2x} + 1}$$

Letting $\beta = \tanh x \implies x = \tanh^{-1}\beta$:
$$\beta (e^{2x} + 1) = e^{2x} - 1$$
$$e^{2x}(1 - \beta) = 1 + \beta \implies e^{2x} = \frac{1 + \beta}{1 - \beta}$$

Taking the natural logarithm:
$$2x = \ln \left( \frac{1 + \beta}{1 - \beta} \right) \implies x = \tanh^{-1}\beta = \frac{1}{2} \ln \left( \frac{1 + \beta}{1 - \beta} \right)$$

Negating both sides using logarithm properties ($\ln(a/b) = -\ln(b/a)$):
$$-\tanh^{-1}\beta = -\frac{1}{2} \ln \left( \frac{1 + \beta}{1 - \beta} \right) = \frac{1}{2} \ln \left( \frac{1 - \beta}{1 + \beta} \right)$$

Substituting this back into the expression for $y'$:
$$y' = y - \tanh^{-1}\beta \quad \blacksquare$$

---

## Part b) Boost Invariance of $\Delta y$ and Collider Relevance

### 1. Invariance of $\Delta y$
For two particles with rapidities $y_1$ and $y_2$, each transforms under a longitudinal boost $\beta$ as:
$$y_1' = y_1 - \tanh^{-1}\beta$$
$$y_2' = y_2 - \tanh^{-1}\beta$$

Their difference in the boosted frame is:
$$\Delta y' = y_1' - y_2' = (y_1 - \tanh^{-1}\beta) - (y_2 - \tanh^{-1}\beta) = y_1 - y_2 = \Delta y$$

Thus, the rapidity difference $\Delta y$ is **Lorentz invariant** under boosts along the beam axis ($z$).

### 2. Relevance at Hadron Colliders
- **Parton Frame Boost:** Protons are composite objects; colliding partons (quarks/gluons) carry arbitrary fractions of proton momentum ($x_1 \neq x_2$), as a result, the center-of-mass frame of the parton-parton interaction is almost always boosted along the beam axis ($z$-axis) relative to the laboratory frame with an unknown velocity:
  $$\beta_z = \frac{x_1 - x_2}{x_1 + x_2}$$.
- **Invariant Separation ($\Delta R$):** The rapidity difference $\Delta y$ (along with azimuthal angle difference $\Delta \phi$) between two particles or jets is identical in both the parton center-of-mass frame and the detector lab frame:
  $$\Delta R = \sqrt{(\Delta y)^2 + (\Delta \phi)^2}$$
  
  This makes jet clustering algorithms and particle correlation measurements universally valid across all collision events.
 
- **Particle Production Shape ($dN/dy$):** Since $dy' = dy$, the particle density distribution $dN/dy$ preserves its shape across events, simply shifting additively along the $y$-axis.

## Part c) Pseudorapidity $\eta$ and the Ultra-Relativistic Limit

The pseudorapidity $\eta$ is defined as[cite: 10]:
$$\eta \equiv -\ln\left(\tan\frac{\theta}{2}\right)$$

where $\theta$ is the polar angle relative to the beam axis ($z$-axis).

---

### 1. Proof that $y \to \eta$ in the Ultra-Relativistic Limit ($|\vec{p}| \gg m$ or $m = 0$)[cite: 10]

The relativistic energy-momentum relation is:
$$E = \sqrt{|\vec{p}|^2 + m^2}$$

In the ultra-relativistic limit $|\vec{p}| \gg m$ (or for massless particles where $m = 0$)[cite: 10]:
$$E \approx |\vec{p}|$$

Using spherical coordinates, the longitudinal momentum along the $z$-axis is[cite: 10]:
$$p_z = |\vec{p}|\cos\theta$$[cite: 10]

Substituting $E \approx |\vec{p}|$ and $p_z = |\vec{p}|\cos\theta$ into the definition of rapidity $y$[cite: 1, 10]:
$$y = \frac{1}{2} \ln \left( \frac{E + p_z}{E - p_z} \right) \approx \frac{1}{2} \ln \left( \frac{|\vec{p}| + |\vec{p}|\cos\theta}{|\vec{p}| - |\vec{p}|\cos\theta} \right) = \frac{1}{2} \ln \left( \frac{1 + \cos\theta}{1 - \cos\theta} \right)$$[cite: 1]

---

### 2. Simplifying the Trigonometric Expression (Using the Hint)

Using the double-angle identities with $x = \frac{\theta}{2}$[cite: 10]:
$$\cos\theta = \cos^2\left(\frac{\theta}{2}\right) - \sin^2\left(\frac{\theta}{2}\right)$$[cite: 10]
$$1 = \cos^2\left(\frac{\theta}{2}\right) + \sin^2\left(\frac{\theta}{2}\right)$$

Adding and subtracting these identities:
1. **Numerator:**
   $$1 + \cos\theta = 2\cos^2\left(\frac{\theta}{2}\right)$$

2. **Denominator:**
   $$1 - \cos\theta = 2\sin^2\left(\frac{\theta}{2}\right)$$

Now take the ratio:
$$\frac{1 + \cos\theta}{1 - \cos\theta} = \frac{2\cos^2(\theta/2)}{2\sin^2(\theta/2)} = \frac{1}{\tan^2(\theta/2)} = \left(\tan\frac{\theta}{2}\right)^{-2}$$

Substitute this back into the logarithm for $y$:
$$y \approx \frac{1}{2} \ln \left[ \left(\tan\frac{\theta}{2}\right)^{-2} \right] = \frac{1}{2} \cdot (-2) \ln\left(\tan\frac{\theta}{2}\right) = -\ln\left(\tan\frac{\theta}{2}\right)$$

Comparing this with the definition of $\eta$[cite: 10]:
$$y \to \eta \quad \blacksquare$$[cite: 10]

---
## Part c) Pseudorapidity $\eta$ and the Ultra-Relativistic Limit

The pseudorapidity $\eta$ is defined as:
$$\eta = -\ln\left(\tan\frac{\theta}{2}\right)$$

**Show that in ultrarelativistic limit ($|\vec{p}| \gg m$, respective for massless particles), it is $y = \eta$:**

When a particle is moving near $\sim c$:
$$E = \sqrt{|\vec{p}|^2 + m^2} \approx |\vec{p}| \quad \text{(mass becomes negligible)}$$

and
$$p_z = |\vec{p}|\cos\theta \quad \text{(momentum component along beam axis)}$$

For rapidity:
$$y = \frac{1}{2} \ln \left( \frac{E + p_z}{E - p_z} \right)$$

$$\frac{E + p_z}{E - p_z} = \frac{|\vec{p}| + |\vec{p}|\cos\theta}{|\vec{p}| - |\vec{p}|\cos\theta} = \frac{1 + \cos\theta}{1 - \cos\theta}$$

---

### Hint: $\cos 2x = \cos^2 x - \sin^2 x$

Taking $x = \frac{\theta}{2}$:
$$\cos\theta = \cos^2\frac{\theta}{2} - \sin^2\frac{\theta}{2}$$

Using $\cos^2\frac{\theta}{2} + \sin^2\frac{\theta}{2} = 1$:
* $\cos\theta = 2\cos^2\frac{\theta}{2} - 1 \implies 1 + \cos\theta = 2\cos^2\frac{\theta}{2}$
* or $\cos\theta = 1 - 2\sin^2\frac{\theta}{2} \implies 1 - \cos\theta = 2\sin^2\frac{\theta}{2}$

Then:
$$\frac{1 + \cos\theta}{1 - \cos\theta} = \frac{\cos^2(\theta/2)}{\sin^2(\theta/2)} = \frac{1}{\tan^2(\theta/2)} = \tan^{-2}\left(\frac{\theta}{2}\right)$$

Substituting into $y$:
$$y = \frac{1}{2} \ln \left( \tan^{-2}\left(\frac{\theta}{2}\right) \right) = -\ln\left(\tan\frac{\theta}{2}\right) = \eta$$

$$\mathbf{y = \eta \quad \text{in ultra-relativistic limit!}}$$

---

**Explain why in measurements at hadron colliders, pseudorapidity is frequently used as an observable rather than $y$:**

* **Particle Identification (PID) not needed:** Calculating $y = \frac{1}{2}\ln\left(\frac{E+p_z}{E-p_z}\right)$ requires knowing the particle's mass $m$ to get $E$. In contrast, $\eta$ only requires measuring the polar angle $\theta$ from detector tracking coordinates, without needing particle identification.
* **Ultra-relativistic regime:** At high-energy hadron colliders, most detected particles have $|\vec{p}| \gg m$, making $\eta \approx y$ an accurate approximation for the boost-invariant properties of rapidity.
### 3. Why Pseudorapidity $\eta$ is Frequently Used Instead of $y$ at Hadron Colliders

1. **Purely Geometric Observable (No Mass/PID Required):**
   - True rapidity $y = \frac{1}{2}\ln\left(\frac{E+p_z}{E-p_z}\right)$ requires knowing both the particle's energy $E$ and momentum $p_z$, which requires knowing its rest mass $m$ (particle identification)[cite: 1].
   - Pseudorapidity $\eta = -\ln(\tan(\theta/2))$ depends **solely on the polar angle $\theta$**[cite: 10]. Detectors can determine $\theta$ immediately from the spatial hit coordinates in tracking systems or calorimeter clusters without needing mass determination.

2. **Ultra-Relativistic Regime at Hadron Colliders:**
   - At high-energy colliders (such as the LHC at $\sqrt{s} = 13\text{-}14\text{ TeV}$), typical particles produced in hard scatterings (pions, kaons, muons, electrons, jet fragments) carry transverse momenta $p_T \gg m$.
   - Because $|\vec{p}| \gg m$ holds for the vast majority of reconstructed particles, $\eta \approx y$ to very high precision[cite: 10].

3. **Detector Geometry and Coverage:**
   - Since $\eta$ maps one-to-one to the physical polar angle $\theta$, experimental detector acceptances (barrel, endcaps, forward calorimeters) are naturally segmented in uniform intervals of $\eta$ and azimuth $\phi$.

# Exercise 2: Higgs-boson decays

A Higgs boson $H$ of mass $M_H$ decays into two photons:
$$H \to \gamma\gamma$$
and moves with velocity $\beta = \frac{p_z}{E}$ along the $+z$ direction relative to the CMS detector.

---
## a) Four-Vectors in the Higgs Rest Frame
Assume in the rest frame of $H$, the momenta of the photons lie in $xz$-plane, and the angle between $\vec{p}_{\gamma_1}'$ and $z$-axis is $\theta'$.
**What are the four-vectors $p_{\gamma_1}^{\prime\mu}$ and $p_{\gamma_2}^{\prime\mu}$ of the two photons in the rest frame of the Higgs boson? State their dependency on $\theta'$.**

Four-momentum notation:
$$p^\mu = \begin{pmatrix} p^0 \\ p^1 \\ p^2 \\ p^3 \end{pmatrix} = \begin{pmatrix} E/c \\ p_x \\ p_y \\ p_z \end{pmatrix} = \begin{pmatrix} E/c \\ \vec{p} \end{pmatrix}$$

In Rest Frame of $H$:
$$P_H^{\prime\mu} = (M_H, 0, 0, 0)$$

$\Rightarrow$ In Four-momentum Conservation:
$$P_H^{\prime\mu} = p_{\gamma_1}^{\prime\mu} + p_{\gamma_2}^{\prime\mu}$$

Since 3-momentum is initially $0$:
$$\vec{p}_{\gamma_1}' = -\vec{p}_{\gamma_2}'$$

and Photons are massless $(m_\gamma = 0)$ so:
$$E_\gamma' = |\vec{p}\,'|$$

$\Rightarrow$ In Energy Conservation:
$$M_H = E_{\gamma_1}' + E_{\gamma_2}'$$
$$\text{Then } E_{\gamma_1}' = E_{\gamma_2}' = \frac{1}{2} M_H = |\vec{p}_{\gamma_1}'| = |\vec{p}_{\gamma_2}'|$$

---

### Spatial components:
* Motion is in $x$-$z$ plane $(p_y' = 0)$
* $\vec{p}_{\gamma_1}'$ makes angle $\theta'$ with $z$-axis

Therefore:
* $p_{z,\gamma_1}' = |\vec{p}\,'| \cos\theta' = \frac{1}{2} M_H \cos\theta'$
* $p_{x,\gamma_1}' = |\vec{p}\,'| \sin\theta' = \frac{1}{2} M_H \sin\theta'$
* $p_{y,\gamma_1}' = 0$

---

### For $2^{\text{nd}}$ photon: $\vec{p}_{\gamma_2}' = -\vec{p}_{\gamma_1}'$
* $p_{z,\gamma_2}' = -\frac{1}{2} M_H \cos\theta'$
* $p_{x,\gamma_2}' = -\frac{1}{2} M_H \sin\theta'$
* $p_{y,\gamma_2}' = 0$

Since $p^{\prime\mu} = (E', p_x', p_y', p_z')^T$:

$$p_{\gamma_1}^{\prime\mu} = \frac{1}{2} M_H \begin{pmatrix} 1 \\ \sin\theta' \\ 0 \\ \cos\theta' \end{pmatrix}$$

$$p_{\gamma_2}^{\prime\mu} = \frac{1}{2} M_H \begin{pmatrix} 1 \\ -\sin\theta' \\ 0 \\ -\cos\theta' \end{pmatrix}$$

## b) Show that photons' four vectors in lab frame, i.e. in the rest frame of CMS detector, are given by:

$$p_{\gamma_1} = \frac{1}{2} M_H \begin{pmatrix} \gamma + \beta\gamma\cos\theta' \\ \sin\theta' \\ 0 \\ \beta\gamma + \gamma\cos\theta' \end{pmatrix} \quad \& \quad p_{\gamma_2} = \frac{1}{2} M_H \begin{pmatrix} \gamma - \beta\gamma\cos\theta' \\ -\sin\theta' \\ 0 \\ \beta\gamma - \gamma\cos\theta' \end{pmatrix}$$

---

Higgs moves in $+z$ dir with velocity $\beta$ relative to CMS detector.  
We need to apply **Lorentz Boost** to transform four-momenta from Higgs rest frame to lab frame (CMS Detector Frame).

### On Standard Transformation (Lab $\to$ Rest Frame):
* Lab sees a particle moving $+ \beta\hat{z}$
* From the perspective of particle's rest frame, lab is rushing backwards at velocity $-\beta\hat{z}$.
* Therefore from lab frame $(t, z) \longrightarrow$ moving frame $(t', z')$ uses $-\beta$:
  $$t' = \gamma (t - \beta z)$$
  $$z' = \gamma (z - \beta t)$$

---

### In this problem: we have the opposite (Rest frame $\to$ Lab Frame)

$$p^{\mu\prime} = (E', p_x', p_y', p_z') \longrightarrow (E, p_x, p_y, p_z)$$

* $E = \gamma (E' + \beta p_z')$
* $p_z = \gamma (p_z' + \beta E')$

*(Rest frame particles carry forward momentum in lab frame!)*

Expressing each component:
* $p^0 = E = \gamma E' + 0 \cdot p_x' + 0 \cdot p_y' + \gamma\beta p_z'$
* $p^1 = p_x = 0 \cdot E' + 1 \cdot p_x' + 0 \cdot p_y' + 0 \cdot p_z'$
* $p^2 = p_y = 0 \cdot E' + 0 \cdot p_x' + 1 \cdot p_y' + 0 \cdot p_z'$
* $p^3 = p_z = \gamma\beta E' + 0 \cdot p_x' + 0 \cdot p_y' + \gamma p_z'$

### Let's put them into Boost matrix:

$$\begin{pmatrix} E \\ p_x \\ p_y \\ p_z \end{pmatrix} = 
\begin{pmatrix} 
\gamma & 0 & 0 & \gamma\beta \\ 
0 & 1 & 0 & 0 \\ 
0 & 0 & 1 & 0 \\ 
\gamma\beta & 0 & 0 & \gamma 
\end{pmatrix} 
\begin{pmatrix} E' \\ p_x' \\ p_y' \\ p_z' \end{pmatrix}$$

---

### For Photon (I):
Where:
* $E_{\gamma_1}' = \frac{1}{2} M_H$
* $p_{x,\gamma_1}' = \frac{1}{2} M_H \sin\theta'$
* $p_{y,\gamma_1}' = 0$
* $p_{z,\gamma_1}' = \frac{1}{2} M_H \cos\theta'$

Applying the transformation:
* $E_{\gamma_1} = \gamma E_1' + \beta\gamma p_{z,1}' = \gamma\left(\frac{1}{2}M_H\right) + \beta\gamma\left(\frac{1}{2}M_H\cos\theta'\right) = \frac{1}{2}M_H (\gamma + \beta\gamma\cos\theta')$
* $p_{x,\gamma_1} = p_{x,\gamma_1}' = \frac{1}{2} M_H \sin\theta'$
* $p_{y,\gamma_1} = p_{y,\gamma_1}' = 0$
* $p_{z,\gamma_1} = \gamma\beta E_1' + \gamma p_{z,1}' = \gamma\beta\left(\frac{1}{2}M_H\right) + \gamma\left(\frac{1}{2}M_H\cos\theta'\right) = \frac{1}{2}M_H (\beta\gamma + \gamma\cos\theta')$

**Then four vectors for Photon 1 becomes in lab frame:**
$$p_{\gamma_1} = \frac{1}{2} M_H \begin{pmatrix} \gamma + \beta\gamma\cos\theta' \\ \sin\theta' \\ 0 \\ \beta\gamma + \gamma\cos\theta' \end{pmatrix}$$

---

### For Photon (II):
Let's apply the boost to Photon II:  
Where:
* $E_{\gamma_2}' = \frac{1}{2} M_H$
* $p_{x,\gamma_2}' = -\frac{1}{2} M_H \sin\theta'$
* $p_{y,\gamma_2}' = 0$
* $p_{z,\gamma_2}' = -\frac{1}{2} M_H \cos\theta'$

Applying the transformation:
* $E_{\gamma_2} = \gamma E_2' + \beta\gamma p_{z,2}' = \gamma\left(\frac{1}{2}M_H\right) + \beta\gamma\left(-\frac{1}{2}M_H\cos\theta'\right) = \frac{1}{2}M_H (\gamma - \beta\gamma\cos\theta')$
* $p_{x,\gamma_2} = p_{x,\gamma_2}' = -\frac{1}{2} M_H \sin\theta'$
* $p_{y,\gamma_2} = p_{y,\gamma_2}' = 0$
* $p_{z,\gamma_2} = \beta\gamma E_2' + \gamma p_{z,2}' = \beta\gamma\left(\frac{1}{2}M_H\right) + \gamma\left(-\frac{1}{2}M_H\cos\theta'\right) = \frac{1}{2}M_H (\beta\gamma - \gamma\cos\theta')$

**Four vectors for Photon II:**
$$p_{\gamma_2} = \frac{1}{2} M_H \begin{pmatrix} \gamma - \beta\gamma\cos\theta' \\ -\sin\theta' \\ 0 \\ \beta\gamma - \gamma\cos\theta' \end{pmatrix} \quad \blacksquare$$

## c) Polar Angle $\theta$ in the Laboratory Frame

From the laboratory frame four-vector of Photon 1:
$$p_{\gamma_1} = \frac{1}{2} M_H \begin{pmatrix} \gamma + \beta\gamma\cos\theta' \\ \sin\theta' \\ 0 \\ \beta\gamma + \gamma\cos\theta' \end{pmatrix}$$

The polar angle $\theta$ with respect to the $z$-axis in the laboratory frame satisfies:
$$\tan\theta = \frac{p_{x,\gamma_1}}{p_{z,\gamma_1}} = \frac{\sin\theta'}{\gamma (\beta + \cos\theta')}$$

Using `math.atan2(y, x)` where $y = p_{x,\gamma_1}$ and $x = p_{z,\gamma_1}$:
$$\theta(\theta') = \text{atan2}\Big(\sin\theta',\, \gamma(\beta + \cos\theta')\Big)$$

---

## d) Angle $\alpha$ Between the Two Photons in the Laboratory Frame

Both photons remain in the $xz$-plane ($p_y = 0$).  
Using the polar angles $\theta_1$ and $\theta_2$ of each photon in the laboratory frame:
$$\theta_1 = \text{atan2}\Big(\sin\theta',\, \gamma(\beta + \cos\theta')\Big)$$
$$\theta_2 = \text{atan2}\Big(-\sin\theta',\, \gamma(\beta - \cos\theta')\Big)$$

The opening angle $\alpha$ between the two photons is:
$$\alpha = |\theta_1 - \theta_2|$$

*(Equivalently, from four-momentum conservation $M_H^2 = 2 E_1 E_2 (1 - \cos\alpha) \implies \cos\alpha = 1 - \frac{M_H^2}{2E_1E_2}$).*

```python
import numpy as np
import matplotlib.pyplot as plt

# Constants
M_H = 125.0  # Higgs mass in GeV
p_z_values = [0, 50, 125, 250, 500]  # Higgs momentum values in GeV

# Array of theta' in the Higgs rest frame from 0 to pi
theta_prime = np.linspace(0, np.pi, 500)

# Create subplots for (c) and (d)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5.5))

for p_z in p_z_values:
    # Kinematics of Higgs
    E_H = np.sqrt(p_z**2 + M_H**2)
    beta = p_z / E_H
    gamma = E_H / M_H
    
    # Photon 1 components (factored out 0.5 * M_H)
    px1 = np.sin(theta_prime)
    pz1 = gamma * (beta + np.cos(theta_prime))
    
    # Photon 2 components
    px2 = -np.sin(theta_prime)
    pz2 = gamma * (beta - np.cos(theta_prime))
    
    # Polar angles in laboratory frame using np.arctan2(y, x)
    theta1 = np.arctan2(px1, pz1)
    theta2 = np.arctan2(px2, pz2)
    
    # Opening angle alpha between the two photon vectors
    # Using 2D dot product between normalized spatial vectors
    dot_prod = (px1 * px2 + pz1 * pz2) / (np.sqrt(px1**2 + pz1**2) * np.sqrt(px2**2 + pz2**2))
    dot_prod = np.clip(dot_prod, -1.0, 1.0)
    alpha = np.arccos(dot_prod)
    
    label = rf"$p_z = {p_z}$ GeV ($\gamma = {gamma:.2f}$)"
    ax1.plot(np.degrees(theta_prime), np.degrees(theta1), label=label, lw=2)
    ax2.plot(np.degrees(theta_prime), np.degrees(alpha), label=label, lw=2)

# Styling Plot (c)
ax1.set_title(r"Part c: Photon 1 Polar Angle $\theta$ vs. $\theta'$", fontsize=13)
ax1.set_xlabel(r"Rest Frame Angle $\theta'$ [degrees]", fontsize=11)
ax1.set_ylabel(r"Lab Frame Angle $\theta$ [degrees]", fontsize=11)
ax1.set_xlim(0, 180)
ax1.set_ylim(0, 180)
ax1.grid(True, linestyle="--", alpha=0.6)
ax1.legend(fontsize=9)

# Styling Plot (d)
ax2.set_title(r"Part d: Opening Angle $\alpha$ Between Photons vs. $\theta'$", fontsize=13)
ax2.set_xlabel(r"Rest Frame Angle $\theta'$ [degrees]", fontsize=11)
ax2.set_ylabel(r"Opening Angle $\alpha$ [degrees]", fontsize=11)
ax2.set_xlim(0, 180)
ax2.set_ylim(0, 190)
ax2.grid(True, linestyle="--", alpha=0.6)
ax2.legend(fontsize=9)

plt.tight_layout()
plt.show()
```
